# 2. Load Dataset

```python
df = pd.read_csv(r"D:\OneDrive\Documents\archive\student_data.csv")
df.head()
```

```python
# Data Understanding 
df.info()
df.describe()
# Data Cleaning 
df.isna().sum()
df.duplicated()
# ----------------------
df.columns 
df.shape
df.duplicated()
```

```python
df.columns
```

```python
# Columns Separation
num_cols = df.select_dtypes(include='int64').columns
num_cols
cat_cols = df.select_dtypes(include='object').columns
cat_cols
```

```python
# Create average marks column
# Featre Engineering 
df['avg_marks'] = (df['G1'] + df['G2'] + df['G3']) / 3
```

# Visualizations 

```python
# Data Analysis 
sns.boxplot(x='sex', y='G3', data=df) 
plt.title("Gender vs Final Marks")
plt.show();
```

```python
n_cols = 3   # number of plots per row
n_rows = (len(cat_cols) // n_cols) + 1
```

```python
for i, col in enumerate(cat_cols):
    if col != 'G3':
        plt.subplot(n_rows, n_cols, i+1)
        sns.boxplot(x=col, y='G3', data=df)
        plt.title(f"{col} vs G3")
        plt.show()
```

```python
n_cols = 3   # number of plots per row
n_rows = (len(num_cols) // n_cols) + 1
```

```python
for i, col in enumerate(num_cols):
    if col != 'G3':
        plt.subplot(n_rows, n_cols, i+1)
        sns.scatterplot(x=col, y='G3', data=df)
        plt.title(f"{col} vs G3")
        plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(df[['G1','G2','G3']].corr(), annot=True)
plt.title("Marks Correlation")
plt.show()
```

```python
sns.histplot(df['avg_marks'])
plt.title("Average Marks Distribution")
plt.show()
```
